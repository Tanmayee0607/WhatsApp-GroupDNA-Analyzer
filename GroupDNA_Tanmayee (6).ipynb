```python
import re

# WhatsApp message pattern
pattern = r"^(\d{2}/\d{2}/\d{2}, \d{2}:\d{2}) - ([^:]+): (.*)$"

# Read the file
with open("DADS Minor PROJECT dataset.txt", "r", encoding="utf-8") as file:
    lines = file.readlines()

messages = []
participants = []

system_messages = 0
media_omitted = 0
deleted_messages = 0

for line in lines:

    line = line.strip()

    match = re.match(pattern, line)

    if match:

        timestamp = match.group(1)
        sender = match.group(2)
        message = match.group(3)

        # Skip media omitted
        if "<Media omitted>" in message:
            media_omitted += 1
            continue

        # Skip deleted messages
        if "This message was deleted" in message or "You deleted this message" in message:
            deleted_messages += 1
            continue

        messages.append({
            "timestamp": timestamp,
            "sender": sender,
            "message": message
        })

        if sender not in participants:
            participants.append(sender)

    else:

        # Multi-line message
        if len(messages) > 0:
            messages[-1]["message"] += " " + line
        else:
            system_messages += 1

# Basic Information
total_messages = len(messages)

start_date = messages[0]["timestamp"].split(", ")[0]
end_date = messages[-1]["timestamp"].split(", ")[0]

print("=" * 50)
print("GROUP OVERVIEW")
print("=" * 50)
print("Total Messages      :", total_messages)
print("Participants        :", len(participants))
print("Period              :", start_date, "to", end_date)
print("System Messages     :", system_messages)
print("Media Omitted       :", media_omitted)
print("Deleted Messages    :", deleted_messages)
print()
print("Participants List")
print("-----------------")

for person in participants:
    print(person)
```

```python
from datetime import datetime

reply_time = {}
reply_count = {}

for i in range(1, len(messages)):

    current = messages[i]
    previous = messages[i-1]

    current_sender = current["sender"]
    previous_sender = previous["sender"]

 # Ignore self replies
    if current_sender != previous_sender:

        current_time = datetime.strptime(current["timestamp"], "%d/%m/%y, %H:%M")
        previous_time = datetime.strptime(previous["timestamp"], "%d/%m/%y, %H:%M")

        difference = (current_time - previous_time).total_seconds()

# Ignoring huge gaps (more than 2 hours)
        if difference <= 7200:

            if current_sender in reply_time:
                reply_time[current_sender] += difference
                reply_count[current_sender] += 1
            else:
                reply_time[current_sender] = difference
                reply_count[current_sender] = 1
```

```python
# ==============================================================================
# PARSING COUNTS (Mapped directly from parser output)
# ==============================================================================
system_messages_count = 4
media_messages_count = 32
deleted_messages_count = 15


# ==============================================================================
# SECTION 1: VISUAL DATA OUTPUTS & GROUP OVERVIEW
# ==============================================================================

print("=" * 72)
print("GROUP DNA REPORT".center(72))
print("HOSTEL BOYS".center(72))
print("=" * 72)
print()

# 1. GROUP OVERVIEW
print("                          GROUP OVERVIEW")
print("-" * 72)
from datetime import datetime

start = datetime.strptime(start_date, "%d/%m/%y")
end = datetime.strptime(end_date, "%d/%m/%y")

total_days = (end - start).days + 1
print(f"Period             : {start_date} to {end_date}")
print(f"Total Duration     : {total_days} Days")
print(f"Total Messages     : {total_messages:,}")
print(f"Active Participants: {len(senders)}")
print(f"System Messages    : {system_messages_count}")
print(f"Media Messages     : {media_messages_count}")
print(f"Deleted Messages   : {deleted_messages_count}")
print()

# ==============================================================================
# MESSAGES PER PERSON (CENTERED HEADER & REORDERED COLUMNS)
# ==============================================================================

print("-" * 72)
print("MESSAGES PER PERSON".center(72))
print("-" * 72)

max_messages = max(message_count.values())

for sender in sorted(message_count, key=message_count.get, reverse=True):

    count = message_count[sender]

    percentage = (count / total_messages) * 100

    bar_length = int((count / max_messages) * 15)

    bar = "█" * bar_length

    print(f"{sender:<8} {bar:<18} {count:>4} ({percentage:.1f}%)")


# ==============================================================================
# ACTIVITY HEATMAP (PURE PYTHON + NUMPY - STRING TIMESTAMP SAFE)
# ==============================================================================

from datetime import datetime

start = datetime.strptime(start_date, "%d/%m/%y")
end = datetime.strptime(end_date, "%d/%m/%y")

total_days = (end - start).days + 1

members_list = list(message_count.keys())
num_members = len(members_list)

# 1. Initialize 2D NumPy array matrix (Members x 24 Hours)
heatmap_matrix = np.zeros((num_members, 24), dtype=int)

# Map member names to matrix row indices
member_to_idx = {name: i for i, name in enumerate(members_list)}

# Helper function to extract integer hour safely from strings or objects
def extract_hour(msg):
    # Check if 'hour' key already exists as int or numeric str
    if 'hour' in msg and msg['hour'] is not None:
        try:
            return int(msg['hour'])
        except ValueError:
            pass

    # Process 'timestamp' / 'datetime' string
    ts = msg.get('timestamp') or msg.get('datetime') or msg.get('date')
    if ts is None:
        return None

    if isinstance(ts, int):
        return ts
    elif hasattr(ts, 'hour'):  # Datetime object
        return ts.hour
    elif isinstance(ts, str):
        # Handle ISO strings like "2026-02-15 14:30:00" or "14:30:00"
        if ' ' in ts:
            time_part = ts.split(' ')[1]
            return int(time_part.split(':')[0])
        elif 'T' in ts:
            time_part = ts.split('T')[1]
            return int(time_part.split(':')[0])
        elif ':' in ts:
            return int(ts.split(':')[0])

    return None

# 2. Populate the 2D matrix directly from raw message list
for msg in messages:
    sender = msg.get('sender') or msg.get('author') or msg.get('name')
    hour = extract_hour(msg)

    if sender in member_to_idx and hour is not None and 0 <= hour < 24:
        row_idx = member_to_idx[sender]
        heatmap_matrix[row_idx, hour] += 1

# 3. Determine max activity for shading thresholds
max_val = heatmap_matrix.max()

def get_shade_symbol(count):
    if count == 0:
        return "."
    ratio = count / max_val if max_val > 0 else 0
    if ratio <= 0.25:
        return "░"
    elif ratio <= 0.50:
        return "▒"
    elif ratio <= 0.75:
        return "▓"
    else:
        return "█"

# 4. Print Heatmap with Exact Formatting
LINE_WIDTH = 90

print("=" * LINE_WIDTH)
print("ACTIVITY HEATMAP".center(LINE_WIDTH))
print("=" * LINE_WIDTH)

# Column Headers (Hours 00 to 23)
hours_header = "".join([f"{h:02d} " for h in range(24)])
print(f"{'Participant':<12}{hours_header}")
print("-" * LINE_WIDTH)

# Print Member Rows
for idx, member in enumerate(members_list):
    row_symbols = "".join([f"{get_shade_symbol(heatmap_matrix[idx, h])}  " for h in range(24)])
    print(f"{member:<12}{row_symbols}")

print("-" * LINE_WIDTH)
print()


print("=" * 90)
print(" "*30 , "THIS GROUP'S FAVOURITE WORDS")
print("=" * 90)

max_count = max(word_count.values())

count = 0

for word in sorted(word_count, key=word_count.get, reverse=True):

    value = word_count[word]

    bar_length = int((value / max_count) * 15)

    bar = "█" * bar_length

    print(f"{word:<10} {bar:<18} {value}")

    count += 1

    if count == 8:
        break
# ==============================================================================
# RESPONSE TIME INSIGHTS (PURE PYTHON - WITH ROBUST TIMESTAMP PARSING)
# ==============================================================================

from datetime import datetime
from collections import defaultdict
import re

LINE_WIDTH = 72

print("=" * LINE_WIDTH)
print("RESPONSE TIME INSIGHTS".center(LINE_WIDTH))
print("=" * LINE_WIDTH)

# Flexible string parser for common chat timestamp formats
def parse_dt(ts):
    if isinstance(ts, datetime):
        return ts
    if not isinstance(ts, str):
        return None

    # Clean non-breaking spaces and clean whitespace
    ts = ts.replace('\u202f', ' ').replace('\xa0', ' ').strip()

    # Try common explicit datetime formats
    formats = [
        "%Y-%m-%d %H:%M:%S",
        "%Y-%m-%dT%H:%M:%S",
        "%d/%m/%Y, %H:%M",
        "%d/%m/%y, %H:%M",
        "%d/%m/%Y, %I:%M %p",
        "%d/%m/%y, %I:%M %p",
        "%m/%d/%Y, %I:%M %p",
        "%m/%d/%y, %I:%M %p",
        "%Y/%m/%d %H:%M:%S"
    ]

    for fmt in formats:
        try:
            return datetime.strptime(ts, fmt)
        except ValueError:
            pass

    # Fallback: Extract numbers using regex if standard parsing fails
    # Looking for: [Day/Month/Year, Hour:Minute]
    nums = re.findall(r'\d+', ts)
    if len(nums) >= 5:
        try:
            # Handle 2-digit vs 4-digit years
            year = int(nums[2])
            if year < 100:
                year += 2000
            return datetime(year, int(nums[1]), int(nums[0]), int(nums[3]), int(nums[4]))
        except Exception:
            pass

    return None

# 1. Collect reply times between consecutive messages from different senders
user_response_gaps = defaultdict(list)
sorted_messages = []

for msg in messages:
    ts = msg.get('timestamp') or msg.get('datetime') or msg.get('date') or msg.get('time')
    dt = parse_dt(ts)
    sender = msg.get('sender') or msg.get('author') or msg.get('name')
    if dt and sender:
        sorted_messages.append((dt, sender))

# Sort messages chronologically
sorted_messages.sort(key=lambda x: x[0])

# 2. Calculate response gaps
for i in range(1, len(sorted_messages)):
    prev_dt, prev_sender = sorted_messages[i-1]
    curr_dt, curr_sender = sorted_messages[i]

    if prev_sender != curr_sender:
        gap_minutes = (curr_dt - prev_dt).total_seconds() / 60.0
        # Ignore huge gaps (> 12 hours) representing new conversations/overnight
        if 0 < gap_minutes <= 720:
            user_response_gaps[curr_sender].append(gap_minutes)

# 3. Compute averages & display
avg_reply_times = {}
for user, gaps in user_response_gaps.items():
    if gaps:
        avg_reply_times[user] = sum(gaps) / len(gaps)

if avg_reply_times:
    fastest_user = min(avg_reply_times, key=avg_reply_times.get)
    slowest_user = max(avg_reply_times, key=avg_reply_times.get)

    fastest_mins = round(avg_reply_times[fastest_user], 1)
    slowest_mins = round(avg_reply_times[slowest_user], 1)

    print(f"Fastest Responder : {fastest_user:<15} ({fastest_mins} mins average)")
    print(f"Slowest Replier   : {slowest_user:<15} ({slowest_mins} mins average)")
else:
    # Debug print to help identify exact string structure if it still fails
    sample_ts = messages[0].get('timestamp') or messages[0].get('datetime') if messages else 'N/A'
    print(f"Could not parse timestamps. Sample timestamp found: '{sample_ts}'")


print()

# ==============================================================================
# PEAK ACTIVITY HIGHLIGHTS (PURE PYTHON - NO PANDAS)
# ==============================================================================

from collections import Counter
from datetime import datetime
import re

LINE_WIDTH = 72

print("=" * LINE_WIDTH)
print("PEAK ACTIVITY HIGHLIGHTS".center(LINE_WIDTH))
print("=" * LINE_WIDTH)

# Helper function to parse timestamps safely into datetime objects
def parse_dt(ts):
    if isinstance(ts, datetime):
        return ts
    if not isinstance(ts, str):
        return None

    ts = ts.replace('\u202f', ' ').replace('\xa0', ' ').strip()

    formats = [
        "%Y-%m-%d %H:%M:%S",
        "%Y-%m-%dT%H:%M:%S",
        "%d/%m/%Y, %H:%M",
        "%d/%m/%y, %H:%M",
        "%d/%m/%Y, %I:%M %p",
        "%d/%m/%y, %I:%M %p",
        "%m/%d/%Y, %I:%M %p",
        "%m/%d/%y, %I:%M %p",
        "%Y/%m/%d %H:%M:%S"
    ]

    for fmt in formats:
        try:
            return datetime.strptime(ts, fmt)
        except ValueError:
            pass

    nums = re.findall(r'\d+', ts)
    if len(nums) >= 5:
        try:
            year = int(nums[2])
            if year < 100:
                year += 2000
            return datetime(year, int(nums[1]), int(nums[0]), int(nums[3]), int(nums[4]))
        except Exception:
            pass

    return None

# 1. Track counts per date and per hour
daily_counts = Counter()
hourly_counts = Counter()

for msg in messages:
    ts = msg.get('timestamp') or msg.get('datetime') or msg.get('date') or msg.get('time')
    dt = parse_dt(ts)

    if dt:
        date_str = dt.strftime("%Y-%m-%d (%A)")  # e.g., "2026-02-15 (Sunday)"
        hour_int = dt.hour

        daily_counts[date_str] += 1
        hourly_counts[hour_int] += 1

# 2. Extract busiest day and busiest hour
if daily_counts and hourly_counts:
    busiest_day, max_day_count = daily_counts.most_common(1)[0]
    busiest_hour_24, max_hour_count = hourly_counts.most_common(1)[0]

    # Format hour display range (e.g., 14 -> "14:00 - 15:00")
    start_hour = f"{busiest_hour_24:02d}:00"
    end_hour = f"{(busiest_hour_24 + 1) % 24:02d}:00"
    hour_range_str = f"{start_hour} - {end_hour}"

    print(f"Busiest Day  : {busiest_day:<28} ({max_day_count} messages)")
    print(f"Busiest Hour : {hour_range_str:<28} ({max_hour_count} messages)")
else:
    sample_ts = messages[0].get('timestamp') or messages[0].get('datetime') if messages else 'N/A'
    print(f"Could not parse timestamps. Sample timestamp found: '{sample_ts}'")


print()

from datetime import datetime

# Store unique active dates for each participant
active_dates = {}

for chat in messages:

    sender = chat["sender"]

    date = datetime.strptime(chat["timestamp"], "%d/%m/%y, %H:%M").date()

    if sender not in active_dates:
        active_dates[sender] = []

    if date not in active_dates[sender]:
        active_dates[sender].append(date)


silent_streak = {}

for sender in active_dates:

    dates = sorted(active_dates[sender])

    longest = 0

    for i in range(1, len(dates)):

        gap = (dates[i] - dates[i-1]).days - 1

        if gap > longest:
            longest = gap

    silent_streak[sender] = longest


print("=" * 72)
print(" "*18 ,  "LONGEST SILENT STREAK")
print("=" * 72)

for sender in sorted(silent_streak, key=silent_streak.get, reverse=True):

    print(f"{sender:<10} : {silent_streak[sender]} days")

print("=" * 72)
print(" "*18, "GROUP DNA ARCHETYPES")
print("=" * 72)

archetypes = [
    ("SPAMMER", spammer, f"{message_count[spammer]} messages"),
    ("GROUP MOTHER", mother, f"{mother_count[mother]} caring messages"),
    ("NIGHT OWL", night_owl, f"{night_count[night_owl]} late messages"),
    ("QUESTION MASTER", question_master, f"{question_count[question_master]} questions"),
    ("STORY TELLER", story_teller, f"{story_count[story_teller]} long messages"),
    ("DRAMA QUEEN", drama_queen, f"{drama_count[drama_queen]} drama words"),
    ("COMEDIAN", comedian, f"{comedy_count[comedian]} funny messages"),
    ("DENIED ALL", denied_all, f"{denial_count[denied_all]} refusals"),
    ("GHOST", ghost, f"{message_count[ghost]} messages"),

]
for role, person, stat in archetypes:
    print(f"{role:<20} : {person:<8} ({stat})")

print("=" * 72)

print(" "*18 , "BONUS ANALYTICS")
print("=" * 72)

print(f"ASSIGNMENT SAVIOR    : ",assignment_master)
print("NUMBER OF REMAINDERS = ",assignment_count[assignment_master])

print("_"*72)

print(" "*18 , "Generated by Tanmayee")
print(" "*16 , "Built with Python + NumPy")

print("_"*72)

```
